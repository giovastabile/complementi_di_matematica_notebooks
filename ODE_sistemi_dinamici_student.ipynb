## 6. Oscillatore di Van der Pol

$$
\ddot x-\mu(1-x^2)\dot x+x=0.
$$

In forma del primo ordine:
$$
\dot x=y, \qquad
\dot y=\mu(1-x^2)y-x.
$$


```python
def van_der_pol(mu):
    def f(t, z):
        x, y = z
        return np.array([y, mu*(1-x**2)*y - x])
    return f

# TODO: integrare con RK4 per mu=1 e mu=20.
# Provare a capire perché il caso mu=20 è molto più difficile per un metodo esplicito.
```
